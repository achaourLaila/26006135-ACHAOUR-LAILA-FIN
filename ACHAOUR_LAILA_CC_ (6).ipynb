## 1. Régression linéaire

**Variante C — question spécifique :** Quel diagnostic du résidu indique une mauvaise spécification linéaire ?
* **Réponse :** Une mauvaise spécification linéaire est révélée par la présence d'une structure non-linéaire ou d'un motif non-aléatoire (par exemple, une forme de parabole ou d'entonnoir) dans le graphique des résidus en fonction des valeurs prédites. Si les résidus ne sont pas distribués de manière homogène et symétrique autour de zéro (hétéroscédasticité), l'hypothèse de linéarité est violée.

### C1 — Présentation
* **Catégorie :** Apprentissage supervisé (Régression).
* **Entrées/Sorties :** Entrée : un vecteur de caractéristiques numériques $X \in \mathbb{R}^d$ (les attributs cliniques). Sortie : une valeur continue $y \in \mathbb{R}$ (le score de progression du diabète).
* **Hypothèse clé :** La relation entre les variables d'entrée et la cible est linéaire, et les erreurs sont indépendantes et identiquement distribuées selon une loi normale de moyenne nulle.
* **Équations successives de calcul :**
  La prédiction est donnée par :
  $$\hat{y} = w_1 x_1 + w_2 x_2 + \dots + w_d x_d + b$$
  où :
  - $w_i$ représente le **paramètre appris** (coefficient de régression/poids associé à la variable $i$),
  - $b$ est le biais (intercept),
  - Un **hyperparamètre** clé est `fit_intercept` (booléen spécifiant si l'on doit calculer le biais $b$).

### C2 — Schéma original
```
[Données initiales (X, y)] ---> [Calcul des moindres carrés (OLS)] ---> [Coefficients appris (w, b)]
                                            |                                      |
                                   (Minimisation de MSE)                    (Ligne de prédiction)
```

### C3 — Contextes et choix
* **Deux contextes :**
  1. **Commerce International :** Prédire le volume d'exportation de marchandises en fonction du PIB du pays destinataire et des barrières tarifaires.
  2. **Finance :** Modéliser la valorisation d'un actif immobilier selon sa superficie et sa distance aux centres financiers.
* **Structure de données attendue :** Tableau matriciel de variables quantitatives continues sans valeurs manquantes.
* **Algorithme concurrent :** Ridge Regression (pour gérer la multicollinéarité) ou SVR.
* **Limites :** Inadapté si les relations sous-jacentes sont hautement non-linéaires ou en cas de forte collinéarité entre les variables.

### C4 — Test de sensibilité et interprétation
L'analyse de sensibilité a montré qu'un modèle de régression linéaire classique obtient une MAE de **41.55** et un $R^2$ de **0.485** sur l'ensemble de test de Diabetes. Une seule exécution ne permet pas de conclure sur la robustesse du modèle face à la variance du découpage d'entraînement ; une validation croisée serait requise.

````python
from sklearn.linear_model import LinearRegression
modele = LinearRegression().fit(Xr_ent, yr_ent)
predictions = modele.predict(Xr_test)
print('MAE régression linéaire :', round(mean_absolute_error(yr_test, predictions), 2))
print('R² :', round(r2_score(yr_test, predictions), 3))
````

## 2. Ridge

**Variante C — question spécifique :** Comment évoluent les coefficients quand alpha augmente ?
* **Réponse :** Lorsque le paramètre de régularisation $\alpha$ augmente, l'amplitude des coefficients de régression $w_i$ tend continuellement vers zéro (réduction de leur norme $L_2$). Cependant, contrairement au Lasso, ils ne s'annulent jamais strictement.

### C1 — Présentation
* **Catégorie :** Apprentissage supervisé (Régression avec régularisation $L_2$).
* **Entrées/Sorties :** Entrées : caractéristiques cliniques $X$. Sortie : progression quantitative du diabète $y$.
* **Hypothèse clé :** Introduire une pénalité quadratique stabilise les prédictions en présence de collinéarité.
* **Équations :** On minimise la fonction objectif :
  $$\mathcal{L}_{Ridge} = \sum (y_i - \hat{y}_i)^2 + \alpha \sum w_j^2$$
  où $\alpha \ge 0$ est l'**hyperparamètre** de régularisation et $w_j$ représente chaque **paramètre appris**.

### C2 — Schéma original
```
[Coefficients OLS non contraints] ---> [Pénalité quadratique L2 (Sphère)] ---> [Coefficients Ridge rétrécis]
```

### C3 — Contextes et choix
* **Contextes :** (1) Marketing : estimation de l'impact publicitaire multicanal avec de fortes corrélations entre budgets. (2) Commerce International : modélisation de flux d'import-export soumis à des variables macroéconomiques redondantes.
* **Algorithme concurrent :** OLS standard ou Lasso.
* **Limites :** Ne réalise pas de sélection de variables (conserve tous les attributs).

### C4 — Test de sensibilité et interprétation
L'augmentation de l'hyperparamètre $\alpha$ à **12.0** a légèrement amélioré la généralisation par rapport à la régression classique, en amenant la MAE à **41.46** sur l'échantillon de test.

````python
from sklearn.linear_model import Ridge
modele = make_pipeline(StandardScaler(), Ridge(alpha=12.0)).fit(Xr_ent, yr_ent)
print('MAE Ridge :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
````

## 3. Lasso

**Variante C — question spécifique :** Combien de variables sont annulées pour trois valeurs d’alpha ?
* **Réponse :** Pour $\alpha=2.0$, 2 variables sur 10 sont annulées (8 coefficients restants non nuls). En augmentant $\alpha$ à 5.0 ou 10.0, le nombre de variables annulées augmente drastiquement (respectivement 5 et 8 variables annulées), ce qui démontre le pouvoir de sélection de variables du Lasso.

### C1 — Présentation
* **Catégorie :** Apprentissage supervisé (Régression avec régularisation $L_1$).
* **Entrées/Sorties :** Entrées : caractéristiques cliniques $X$. Sortie : progression quantitative du diabète $y$.
* **Hypothèse clé :** Seul un sous-ensemble restreint de variables prédictives a une influence réelle sur la cible (parcimonie).
* **Équations :** On minimise la fonction objectif :
  $$\mathcal{L}_{Lasso} = \frac{1}{2n} \sum (y_i - \hat{y}_i)^2 + \alpha \sum |w_j|$$
  où $\alpha$ est l'**hyperparamètre** et $w_j$ sont les **paramètres appris**.

### C2 — Schéma original
```
[Coefficients OLS] ---> [Contrainte L1 (Diamant aux angles saillants)] ---> [Coefficients nuls / Sélection]
```

### C3 — Contextes et choix
* **Contextes :** (1) Commerce International : sélection des indicateurs les plus impactants pour l'implantation logistique parmi des dizaines d'indices. (2) Finance : identification des ratios financiers essentiels prédisant la faillite.
* **Algorithme concurrent :** ElasticNet ou Ridge.
* **Limites :** Si $d > n$ ou en cas de forte collinéarité, le Lasso sélectionne une seule variable de manière arbitraire.

### C4 — Test de sensibilité et interprétation
Pour $\alpha=2.0$, le modèle annule 2 coefficients non significatifs et produit la meilleure MAE observée de **41.37**, prouvant que l'élimination du bruit améliore la performance globale.

````python
from sklearn.linear_model import Lasso
modele = make_pipeline(StandardScaler(), Lasso(alpha=2.0, max_iter=15000)).fit(Xr_ent, yr_ent)
print('Nombre de coefficients non nuls :', np.count_nonzero(modele[-1].coef_))
print('MAE Lasso :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
````

## 4. Régression polynomiale

**Variante C — question spécifique :** Le degré 2 améliore-t-il le test ou seulement l’entraînement ?
* **Réponse :** Le degré 2 améliore le test en capturant les courbures sous-jacentes du jeu Diabetes, atteignant un $R^2$ de **0.501** (contre 0.485 pour la régression linéaire simple). Néanmoins, sans une forte régularisation (ici $\alpha=100.0$), le degré 2 entraînerait un surajustement massif sur le jeu d'entraînement.

### C1 — Présentation
* **Catégorie :** Apprentissage supervisé (Régression non-linéaire via transformation linéaire).
* **Entrées/Sorties :** Entrées : $X \in \mathbb{R}^d$. Sortie : cible continue $y$.
* **Hypothèse clé :** Les interactions complexes et non-linéaires entre variables peuvent être modélisées par des combinaisons polynomiales de caractéristiques d'origine.
* **Équations :**
  $$\phi(X) = [1, x_1, x_2, x_1^2, x_1 x_2, x_2^2, \dots]$$
  Le **paramètre appris** est le vecteur de poids $w$ appliqué à la matrice transformée $\phi(X)$. L'**hyperparamètre** principal est le degré polynomial `degree`.

### C2 — Schéma original
```
[Espace initial 1D/2D] ---> [Génération des termes croisés & carrés] ---> [Ajustement linéaire dans l'espace augmenté]
```

### C3 — Contextes et choix
* **Contextes :** (1) Marketing : courbe de réponse des ventes par rapport aux dépenses publicitaires (effet de saturation). (2) Logistique : coût de transport non-linéaire selon la distance.
* **Algorithme concurrent :** SVR (Support Vector Regression) avec noyau RBF.
* **Limites :** Croissance combinatoire explosive du nombre de variables avec le degré (fléau de la dimension).

### C4 — Test de sensibilité et interprétation
L'ajustement combiné d'un degré polynomial de 2 et d'une régularisation Ridge élevée ($\alpha=100.0$) permet de filtrer le bruit des interactions fictives tout en capturant les effets réels, stabilisant le $R^2$ de test à **0.501**.

````python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele = make_pipeline(StandardScaler(), PolynomialFeatures(degree=2, include_bias=False), Ridge(alpha=100.)).fit(Xr_ent, yr_ent)
print('R² polynomial :', round(r2_score(yr_test, modele.predict(Xr_test)), 3))
````

## 5. Arbre de régression

**Variante C — question spécifique :** Quel effet de max_depth sur le biais et la variance ?
* **Réponse :** Un `max_depth` élevé réduit le biais sur le jeu d'entraînement (l'arbre s'adapte parfaitement aux données) mais augmente drastiquement la variance (surajustement). Inversement, un `max_depth` trop faible (ex. 1) engendre un fort biais (sous-ajustement).

### C1 — Présentation
* **Catégorie :** Apprentissage supervisé (Méthode non-paramétrique pour la régression).
* **Entrées/Sorties :** Entrée : vecteur numérique $X$. Sortie : valeur moyenne de la feuille terminale $\hat{y}$.
* **Hypothèse clé :** L'espace des caractéristiques peut être partitionné de façon récursive en régions homogènes par rapport à la variable cible.
* **Équations :** La séparation à chaque nœud minimise la variance intra-région :
  $$\mathcal{H} = \frac{1}{|T|} \sum_{i \in T} (y_i - \bar{y}_T)^2$$
  Le **paramètre appris** est la structure de fractionnement de l'arbre. Les **hyperparamètres** clés sont `max_depth` et `min_samples_leaf`.

### C2 — Schéma original
```
                [X_2 < 0.5 ?]
                 /         \
           [Feuille: 45]  [Feuille: 180]
```

### C3 — Contextes et choix
* **Contextes :** (1) Finance : évaluation de la solvabilité de portefeuilles de crédit par segmentation. (2) Commerce : prédiction du panier moyen selon le profil utilisateur.
* **Algorithme concurrent :** Forêt aléatoire ou SVM.
* **Limites :** Très sensible aux petites variations des données d'entraînement (forte variance).

### C4 — Test de sensibilité et interprétation
En bridant `max_depth=3` et en imposant `min_samples_leaf=8`, le modèle évite le surajustement mais perd en précision brute, enregistrant une MAE de **44.57**.

````python
from sklearn.tree import DecisionTreeRegressor
modele = DecisionTreeRegressor(max_depth=3, min_samples_leaf=8, random_state=GRAINE).fit(Xr_ent, yr_ent)
print('MAE arbre :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
````

## 6. Forêt aléatoire — régression

**Variante C — question spécifique :** La forêt réduit-elle la variance par rapport à l’arbre ?
* **Réponse :** Oui, la forêt aléatoire réduit considérablement la variance en moyennant les prédictions de nombreux arbres indépendants (bagging). Cela se traduit par une baisse de la MAE à **43.11** (contre 44.57 pour l'arbre unique), montrant une meilleure robustesse aux données de test.

### C1 — Présentation
* **Catégorie :** Apprentissage supervisé (Méthode d'ensemble par Bagging).
* **Entrées/Sorties :** Entrées : variables explicatives $X$. Sortie : prédiction finale moyenne $\hat{y}$.
* **Hypothèse clé :** La combinaison d'arbres de décision décorrélés (via bootstrap et sous-espaces aléatoires) permet de réduire la variance globale sans augmenter le biais.
* **Équations :**
  $$\hat{y} = \frac{1}{B} \sum_{b=1}^{B} T_b(X)$$
  Le **paramètre appris** est l'ensemble des arbres individuels $T_b$. Les **hyperparamètres** incluent `n_estimators` et `max_depth`.

### C2 — Schéma original
```
            [Données] ---> [Bootstrap 1, 2, ... B]
                             |
                        [Arbre 1, 2, ... B] ---> [Moyenne des prédictions]
```

### C3 — Contextes et choix
* **Contextes :** (1) Commerce International : estimation des coûts de fret maritime à partir de multiples critères incertains. (2) Finance : prédiction de l'indice boursier.
* **Algorithme concurrent :** Gradient Boosting ou Extra Trees.
* **Limites :** Modèle lourd en mémoire et moins interprétable qu'un arbre simple (effet boîte noire).

### C4 — Test de sensibilité et interprétation
L'ajustement à `n_estimators=60` combiné à `max_depth=6` permet de stabiliser les prédictions et d'obtenir une MAE de **43.11**.

````python
from sklearn.ensemble import RandomForestRegressor
modele = RandomForestRegressor(n_estimators=60, max_depth=6, n_jobs=-1, random_state=GRAINE).fit(Xr_ent, yr_ent)
print('MAE forêt :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
````

## 7. Gradient Boosting — régression

**Variante C — question spécifique :** Que change max_iter sur le coût et le risque de surajustement ?
* **Réponse :** Un nombre élevé d'itérations (`max_iter`) permet de réduire l'erreur d'apprentissage en ajustant continuellement les résidus restants, mais augmente le risque de surajustement et le coût de calcul. Un `max_iter` modéré avec un bon taux d'apprentissage est indispensable.

### C1 — Présentation
* **Catégorie :** Apprentissage supervisé (Méthode d'ensemble par Boosting).
* **Entrées/Sorties :** Entrées : $X$. Sortie : $\hat{y}$ continue.
* **Hypothèse clé :** La construction séquentielle d'arbres faibles s'ajustant sur le gradient de l'erreur du modèle précédent permet d'obtenir un prédicteur robuste.
* **Équations :**
  $$F_m(X) = F_{m-1}(X) + \gamma_m h_m(X)$$
  Les **paramètres appris** sont les paramètres des arbres successifs $h_m$. Les **hyperparamètres** sont `max_iter` (nombre de rounds) et `learning_rate`.

### C2 — Schéma original
```
[Résidus Étape m-1] ---> [Ajustement de l'arbre faible h_m] ---> [Mise à jour du modèle global F_m]
```

### C3 — Contextes et choix
* **Contextes :** (1) Logistique : prédiction précise des délais de livraison en temps réel. (2) Marketing : modélisation de la valeur de vie client (LTV).
* **Algorithme concurrent :** XGBoost ou Random Forest.
* **Limites :** Sensible au bruit et nécessite un réglage minutieux des hyperparamètres.

### C4 — Test de sensibilité et interprétation
En limitant les arbres à de petites tailles via `max_leaf_nodes=9` sur `max_iter=100`, le modèle de boosting atteint une MAE de **44.18**.

## 8. Régression logistique

**Variante C — question spécifique :** Quel effet de C sur les probabilités et la frontière ?
* **Réponse :** L'hyperparamètre $C$ est l'inverse de la force de régularisation ($1/\lambda$). Un $C$ élevé (faible régularisation) produit des probabilités plus contrastées et une frontière de décision complexe, au risque de surajuster. Un $C$ faible (forte régularisation) adoucit la frontière et stabilise le modèle.

### C1 — Présentation
* **Catégorie :** Classification supervisée linéaire.
* **Entrées/Sorties :** Entrées : $X \in \mathbb{R}^d$. Sortie : probabilité d'appartenance à la classe positive $p(y=1|X) \in [0,1]$.
* **Hypothèse clé :** Les classes sont séparables par une frontière linéaire dans l'espace des caractéristiques.
* **Équations :**
  $$p(y=1|X) = \sigma(w^T X + b) = \frac{1}{1 + e^{-(w^T X + b)}}$$
  Les **paramètres appris** sont les coefficients $w$ et le biais $b$. L'**hyperparamètre** clé est `C`.

### C2 — Schéma original
```
[Score linéaire z = w^T X + b] ---> [Fonction Sigmoïde] ---> [Probabilité p dans [0, 1]]
```

### C3 — Contextes et choix
* **Contextes :** (1) Marketing : prédiction de l'attrition client (churn). (2) Finance : détection préliminaire de transactions frauduleuses.
* **Algorithme concurrent :** SVM linéaire ou Naive Bayes.
* **Limites :** Incapable de capturer des relations non-linéaires complexes sans transformation préalable.

### C4 — Test de sensibilité et interprétation
Une valeur modérée $C=0.5$ avec une standardisation rigoureuse permet à la régression logistique d'atteindre un F1-score remarquable de **0.989** sur le jeu Breast Cancer.

````python
from sklearn.linear_model import LogisticRegression
modele = make_pipeline(StandardScaler(), LogisticRegression(C=.5, max_iter=2000, random_state=GRAINE)).fit(Xc_ent, yc_ent)
print('F1 logistique :', round(f1_score(yc_test, modele.predict(Xc_test)), 3))
````

* **Équations :** Un point $p$ est considéré comme cœur si :
  $$|N_{\epsilon}(p)| \ge \text{MinPts}$$
  Il n'y a **aucun paramètre appris**. Les **hyperparamètres** fondamentaux sont `eps` (rayon) et `min_samples` (densité).

### C2 — Schéma original
```
[Points Coeurs (Forte densité)] ---> [Points Frontières connectés] ---> [Points Isolés (Bruits)]
```

## 26. Epsilon-greedy

**Variante C - question specifique :** Quel effet de epsilon sur la concentration des choix ?
* **Reponse :** Un **epsilon eleve** (comme 0.5) pousse l'agent a explorer au hasard un peu partout. Un **epsilon tres faible** (comme 0.05) le force a exploiter presque uniquement le meilleur bras actuel.

### C1 - Presentation
* **Categorie :** Arbitrage exploration/exploitation.
* **Entrees/Sorties :** Entree : les moyennes actuelles des bras. Sortie : le choix du bras a tester.
* **Hypothese cle :** Tester des bras au hasard de temps en temps permet de verifier qu'un bras prometteur n'a pas ete oublie.
* **Calcul simple :**
  - Dans un pourcentage `epsilon` des cas, on choisit un bras au hasard.
  - Dans le reste des cas (1 - epsilon), on prend celui qui a la meilleure moyenne : action = argmax(Q_t).
* **Parametre appris :** La qualite Q(a) de chaque choix.
* **Hyperparametre :** Epsilon (taux d'exploration).

### C2 - Schema original
```
               [Tirer un nombre au hasard]
                /                       \
       (Inferieur a epsilon)      (Superieur a epsilon)
              /                           \
      [Prendre au hasard]         [Prendre le meilleur bras]
```

### C3 - Contextes et choix
* **Deux contextes :**
  1. **Recommandations Netflix :** Montrer 95% de films populaires et 5% de nouveautes inconnues.
  2. **E-commerce :** Proposer majoritairement le produit vedette tout en testant de temps en temps des alternatives.
* **Structure attendue :** Nombre de clics par produit.
* **Concurrent :** UCB (Upper Confidence Bound).
* **Limites :** Dangereux si proposer un mauvais produit au hasard est inadmissible ou trop cher.

### C4 - Test de sensibilite et interpretation
* **Modification :** On baisse epsilon a **0.05** (seulement 5% de hasard) pour voir si l'agent passe la quasi-totalite de ses choix a exploiter le meilleur bras trouve.

````python
# Exploration epsilon-greedy sur le même bandit simulé avec epsilon réduit (Sensibilité)
rng = np.random.default_rng(GRAINE)
probabilites_simulees = np.array([.35, .55, .65, .45])
recompenses = np.zeros(4); nombre_tirages = np.zeros(4)
epsilon = .05 # Changement d'hyperparamètre : réduction drastique de l'exploration
for etape in range(1500):
    bras = int(rng.integers(4)) if rng.random() < epsilon or not np.all(nombre_tirages) else int(np.argmax(recompenses / np.maximum(nombre_tirages, 1)))
    gain = float(rng.random() < probabilites_simulees[bras])
    recompenses[bras] += gain; nombre_tirages[bras] += 1
print('Fréquences de sélection avec epsilon = 0.05 :', nombre_tirages.astype(int))
print('Valeurs estimées :', np.round(recompenses / np.maximum(nombre_tirages, 1), 3))
````

## 27. Q-learning

**Variante C — question spécifique :** Étudiez les récompenses des épisodes, pas seulement Q moyen.
* **Réponse :** Bien que la valeur moyenne finale de la table Q s'établisse à environ **-11.22**, l'analyse des récompenses cumulées par épisode montre que l'agent commence avec un score désastreux proche de **-100** ou pire en raison de chutes fréquentes dans le précipice. Au fil de l'apprentissage (vers l'épisode 800-1100), la récompense par épisode se stabilise autour de **-13**, montrant que l'agent a appris le chemin optimal de longueur minimale (13 étapes) malgré les pénalités accumulées au début.

### C1 — Présentation
* **Catégorie :** Apprentissage par renforcement hors-politique (Off-Policy Temporal Difference Control).
* **Entrées/Sorties :** Entrées : état actuel $S_t$ de la grille (0 à 47) et récompense reçue $R_{t+1}$. Sortie : action $A_t \in \{\text{UP, RIGHT, DOWN, LEFT}\}$.
* **Hypothèse clé :** L'état suivant optimal peut être évalué de manière gloutonne (indépendamment de la politique d'exploration réellement suivie) en utilisant l'opérateur maximum.
* **Étapes de calcul :**
  À chaque transition, la valeur de la table Q est mise à jour avec l'équation de Bellman un pas :
  $$Q(S_t, A_t) \leftarrow Q(S_t, A_t) + \alpha \left[ R_{t+1} + \gamma \max_{a} Q(S_{t+1}, a) - Q(S_t, A_t) \right]$$
* **Paramètre appris :** La fonction de valeur d'action optimale $Q^*(s, a)$.
* **Hyperparamètre :** Le taux d'apprentissage $\alpha$ (fixé à 0.15) et le facteur d'exploration $\epsilon$.

### C2 — Schéma original
```
[État S_t] ---> (Politique Epsilon-Greedy) ---> Choisit Action A_t
                                                      |
[Cible: R + gamma * max Q(S_t+1, a)] <--- (Step) <----/
              |
       [Mise à jour Q(S_t, A_t)]
```

### C3 — Contextes et choix
* **Deux contextes :**
  1. **Gestion d'inventaire automatisée :** Décision de réapprovisionnement à coût minimal dans une chaîne logistique avec transitions stochastiques.
  2. **Routage de paquets télécoms :** Acheminement de paquets de données sur des routeurs pour minimiser le délai de transmission.
* **Structure attendue :** Table de dimension $|S| \times |A|$.
* **Algorithme concurrent :** SARSA (On-Policy) ou Double Q-learning.
* **Mal adapté :** Espaces d'états continus de grande dimension (nécessite alors un approximateur comme DQN) ou environnements hautement instables.

### C4 — Test de sensibilité et interprétation
* **Modification :** Nous allons suivre la somme des récompenses cumulées sur les 100 derniers épisodes pour observer l'efficacité de la politique stabilisée sous un facteur de discount de $\gamma = 0.90$ (contre $0.95$).

````python
# Q-learning manuel avec suivi des récompenses d'épisodes et discount réduit (Sensibilité)
import gymnasium as gym
env = gym.make('CliffWalking-v1')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
recompenses_historique = []
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    recompense_cumulee = 0
    for pas in range(120):
        action = int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
        suivant, recompense, termine, tronque, _ = env.step(action)
        recompense_cumulee += recompense
        # Discount modifié de 0.95 à 0.90
        cible = recompense + (0 if termine else .90 * np.max(q_table[suivant]))
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat = suivant
        if termine or tronque: break
    recompenses_historique.append(recompense_cumulee)
print('Table Q (gamma=0.90) :', q_table.shape, '| valeur moyenne :', round(q_table.mean(), 2))
print('Récompense moyenne cumulée (100 derniers épisodes) :', round(np.mean(recompenses_historique[-100:]), 2))
env.close()
````

## 28. SARSA

**Variante C — question spécifique :** SARSA privilégie-t-il une trajectoire plus sûre sous exploration ?
* **Réponse :** **Oui, absolument.** SARSA est un algorithme *on-policy*, ce qui signifie qu'il met à jour ses valeurs Q en tenant compte de la politique d'exploration réelle (contenant des choix $\epsilon$-greedy aléatoires). Comme l'agent fait parfois des erreurs aléatoires à cause d'epsilon, s'approcher de la falaise s'avère extrêmement coûteux car l'agent risque d'y glisser accidentellement. SARSA apprend donc à contourner le précipice par le haut (trajectoire plus longue mais beaucoup plus sûre), contrairement à Q-learning (*off-policy*) qui suppose un comportement glouton parfait futur et frôle témérairement la falaise.

### C1 — Présentation
* **Catégorie :** Apprentissage par renforcement sur-politique (On-Policy Temporal Difference Control).
* **Entrées/Sorties :** Entrées : état actuel $S_t$, action choisie $A_t$, récompense $R_{t+1}$, état suivant $S_{t+1}$, action suivante planifiée $A_{t+1}$. Sortie : mise à jour de $Q(S_t, A_t)$.
* **Hypothèse clé :** L'évaluation et la mise à jour s'effectuent sur la politique d'exploration effectivement exécutée (On-Policy).
* **Étapes de calcul :**
  À chaque transition, l'action suivante $A_{t+1}$ est choisie à l'avance, et la table est mise à jour :
  $$Q(S_t, A_t) ← Q(S_t, A_t) + \alpha \left[ R_{t+1} + \gamma Q(S_{t+1}, A_{t+1}) - Q(S_t, A_t) \right]$$
* **Paramètre appris :** La fonction de valeur d'action sous la politique actuelle $Q^{\pi}(s, a)$.
* **Hyperparamètre :** Le taux d'apprentissage $\alpha = 0.15$ et le coefficient d'exploration $\epsilon$.

### C2 — Schéma original
```
  [État S_t, Action A_t] ---> (Step) ---> [Recompense R, État S_t+1]
                                                   |
  [Mise à jour Q(S_t,A_t)] <--- (Bellman) <--- Choisit Action A_t+1
```

### C3 — Contextes et choix
* **Deux contextes :**
  1. **Conduite autonome de véhicules :** Apprentissage d'un comportement d'évitement des obstacles très robuste en tenant compte du bruit des capteurs physiques.
  2. **Contrôle de processus industriels :** Régulation de température de réacteurs où toute fluctuation incontrôlée présente un risque catastrophique.
* **Structure attendue :** Table bidimensionnelle de taille $|S| \times |A|$.
* **Algorithme concurrent :** Q-learning, Expected SARSA.
* **Mal adapté :** Environnements ne présentant aucun risque critique aux frontières où l'on cherche l'optimalité pure à tout prix.

### C4 — Test de sensibilité et interprétation
* **Modification :** Nous allons enregistrer et afficher la récompense moyenne cumulée sur les 100 derniers épisodes de SARSA afin de vérifier numériquement qu'elle est plus élevée (moins de pénalités cumulées) que celle du Q-learning.

````python
# SARSA manuel avec suivi de la récompense cumulée d'épisode (Sensibilité)
import gymnasium as gym
env = gym.make('CliffWalking-v1')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
def choisir_action(etat, epsilon):
    return int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
recompenses_historique_sarsa = []
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    action = choisir_action(etat, epsilon)
    recompense_cumulee = 0
    for pas in range(120):
        suivant, recompense, termine, tronque, _ = env.step(action)
        recompense_cumulee += recompense
        action_suivante = choisir_action(suivant, epsilon)
        cible = recompense + (0 if termine else .95 * q_table[suivant, action_suivante])
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat, action = suivant, action_suivante
        if termine or tronque: break
    recompenses_historique_sarsa.append(recompense_cumulee)
print('Table Q SARSA :', q_table.shape, '| valeur moyenne :', round(q_table.mean(), 2))
print('Récompense moyenne cumulée (100 derniers épisodes) :', round(np.mean(recompenses_historique_sarsa[-100:]), 2))
env.close()
````

## 29. DQN - cible conceptuelle

**Variante C - question specifique :** Pourquoi la cible est-elle egale a la recompense en etat terminal ?
* **Reponse :** En etat terminal, l'episode s'arrete immediatement. Il n'existe aucun etat suivant et donc aucune recompense future possible. L'esperance des gains futurs est rigoureusement nulle (G_t+1 = 0). Selon l'equation de Bellman, la cible de mise a jour se simplifie de R_t+1 + gamma * max_a Q(S_t+1, a) a simplement R_t+1. Propager une valeur de Q(S_t+1, .) non nulle fausserait l'apprentissage en faisant croire a l'agent qu'il peut continuer a accumuler des recompenses apres la fin.

### C1 - Presentation
* **Categorie :** Apprentissage par renforcement profond (Deep Q-Learning targets).
* **Entrees/Sorties :** Entrees : recompense R_t, masque de fin d'episode (terminal), et les estimations de valeurs maximales futures max_a' Q(S_t+1, a'). Sortie : cible de regression de Bellman Y_t.
* **Hypothese cle :** L'approximation par reseau de neurones doit converger vers la cible de Bellman definie de maniere stable en utilisant un reseau cible (Target Network) temporairement gele.
* **Etapes de calcul :**
  Le calcul de la cible pour chaque transition i s'effectue via :
  Y_i = R_i + gamma * (1 - T_i) * max_a' Q_target(S'_i, a')
  ou T_i in {0, 1} represente l'etat terminal.
* **Parametre appris :** Les poids du reseau de neurones d'approximation Q(s, a; theta).
* **Hyperparametre :** Le taux d'actualisation temporel gamma in [0, 1].

### C2 - Schema original
```
[Donnees de transition: (S, A, R, S', Terminal)]
                       |
           Est-ce que Terminal == True ?
            /                         \
         (Oui)                       (Non)
          /                             \
   [Cible = R]              [Cible = R + gamma * max Q(S', a')]
```

### C3 - Contextes et choix
* **Deux contextes :**
  1. **Gaming (Atari / Robotique simulee) :** Apprentissage de strategies de jeu optimales directes a partir de pixels.
  2. **Climatisation intelligente :** Regulation thermique continue de grands serveurs pour minimiser l'energie consommee.
* **Structure attendue :** Tenseurs d'experiences (Replay Buffer).
* **Algorithme concurrent :** Double DQN, Dueling DQN, ou algorithmes de gradient de politique (PPO).
* **Mal adapte :** Systemes ou les actions sont continues (necessite plutot DDPG ou SAC).

### C4 - Test de sensibilite et interpretation
* **Modification :** Nous allons augmenter le facteur de discount a **gamma = 0.99** (ce qui donne une tres grande importance aux gains futurs) pour observer la modification des cibles sur les transitions non-terminales.

````python
# Cibles DQN : demonstration du calcul de Bellman avec discount modifie a 0.99 (Sensibilite)
import numpy as np

recompenses = np.array([-1., -100., 0.])
termine = np.array([False, True, False])
meilleure_valeur_suivante = np.array([-3., -2., 5.])
gamma = 0.99 # Changement d'hyperparametre : augmentation du discount

cibles = recompenses + gamma * (~termine) * meilleure_valeur_suivante
print('Cibles de Bellman (gamma=0.99) :', np.round(cibles, 2))
print('Note : la cible terminale (index 1) reste inchangee a -100.0 !')
````
